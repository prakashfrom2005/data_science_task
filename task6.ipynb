```python
num_col=df.select_dtypes(include="number").columns
```

```python
df[num_col]=df[num_col].fillna(df[num_col].median())
```

```python
df.isnull().sum()
```

```python
cat_col=df.select_dtypes(include=["object","category"]).columns
```

```python
for col in cat_col:
    if df[col].isnull().any() and not df[col].mode().empty:
        df[col]=df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
if "Date" in df.columns:
    df["date"]=pd.to_datetime(df["Date"],errors="coerce")
```

```python
if "Gender" in df.columns:
    le=LabelEncoder()
    df["G_encoder"]=le.fit_transform(df["Gender"].astype(str))
```

```python
if "Payment_Method" in df.columns:
    df=pd.get_dummies(df,columns=["Payment_Method"],dtype=int)
    print(df.head())
```

```python
if "Membership_Level" in df.columns:
    oe=OrdinalEncoder(
        categories=[["Bronze", "Silver", "Gold", "Premium"]])
```

```python
df["Membership_Level_encoded"]=oe.fit_transform(
    df[["Membership_Level"]])
```
