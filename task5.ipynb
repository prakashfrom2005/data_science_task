```python
df["Date"]=pd.to_datetime(df['Timestamp'])
```

```python
df=df.sort_values("Date")
```

```python
df
```

```python
plt.plot(df["Date"],df["Temperature (°C)"])
plt.xlabel("date")
plt.ylabel("temperature")
plt.title("Date vs temperature")
plt.show()
```

```python
df
```

```python
df["moving_avg"]=df["Temperature (°C)"].rolling(window=10).mean()
```

```python
df
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["Date"],df["moving_avg"],label="10-days moving average")

plt.xlabel("date")
plt.ylabel("temperature")
plt.title("Date vs moving average")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
